# Подключение исследовательского ноутбука к MLflow

Этот ноутбук нужен, чтобы проверить подключение исследовательской среды к локальному MLflow-серверу и S3-совместимому хранилищу артефактов MinIO.

````python
import os

import mlflow
from mlflow.tracking import MlflowClient

os.environ.setdefault("AWS_ACCESS_KEY_ID", "mlflow")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "mlflow-secret")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://127.0.0.1:9000")

mlflow.set_tracking_uri(os.getenv("MLFLOW_TRACKING_URI", "http://127.0.0.1:5000"))
client = MlflowClient()
mlflow.get_tracking_uri()
````

````python
experiments = client.search_experiments()
[(experiment.experiment_id, experiment.name) for experiment in experiments]
````
