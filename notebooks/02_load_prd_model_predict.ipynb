# Загрузка PRD-модели из MLflow

Чистый проверочный ноутбук: подключаемся к локальному MLflow-серверу, загружаем версию модели с alias `PRD` и выполняем тестовый прогноз.

```python
import os

import mlflow
import pandas as pd

os.environ.setdefault("AWS_ACCESS_KEY_ID", "mlflow")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "mlflow-secret")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://127.0.0.1:9000")

mlflow.set_tracking_uri(os.getenv("MLFLOW_TRACKING_URI", "http://127.0.0.1:5000"))
mlflow.get_tracking_uri()
```

```python
registered_model_name = "stock_price_forecaster_prd"
model_uri = f"models:/{registered_model_name}@PRD"

model = mlflow.pyfunc.load_model(model_uri)
model
```

```python
prediction = model.predict(pd.DataFrame({"horizon": [10.0]}))
prediction
```
